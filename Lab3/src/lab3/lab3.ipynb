## utils

```python
from sklearn.metrics import ConfusionMatrixDisplay

POSITIVE_CLASS = 1

def calculate_metrics(model, name, plots=True):
    pred = model.predict(X_test)
    score = model.predict_proba(X_test)[:, POSITIVE_CLASS] if hasattr(model, 'predict_proba') else model.decision_function(X_test)

    if plots:
        fig, ax = plt.subplots(1, 2, figsize=(10, 4))
        ConfusionMatrixDisplay.from_predictions(y_test, pred, ax=ax[0])
        RocCurveDisplay.from_predictions(y_test, score, ax=ax[1])
        ax[0].set_title(f'Confusion matrix ({name})')
        ax[1].set_title(f'ROC ({name})')
        plt.tight_layout()
        plt.show()
        print(classification_report(y_test, pred))

    return pd.Series({
        'Accuracy': accuracy_score(y_test, pred),
        'Precision': precision_score(y_test, pred, zero_division=0),
        'Recall': recall_score(y_test, pred, zero_division=0),
        'F1': f1_score(y_test, pred, zero_division=0),
        'ROC-AUC': roc_auc_score(y_test, score),
    }, name=name)
```

### Выбор целевого признака

В качестве целевого признака будем использовать признак `unlocked`, поскольку он является бинарным и не таким очевидным и тривиальным, как, например, наличие встроенной графики (хотя тоже уместно)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, 
    f1_score, roc_auc_score, confusion_matrix, classification_report,
    RocCurveDisplay
)


SEED = 4323245
TARGET = 'unlocked'
TEST_SIZE = 0.2

df = df.dropna(subset=[TARGET])


X = df.drop(columns=[TARGET])
y = df[TARGET].map({'Yes': 1, 'No': 0}).astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=SEED,
    stratify=y #стратификация по таргету
)

```

```python
train_stats = pd.DataFrame({
    'missing_ratio': X_train.isna().mean() * 100,
    'num_unique': X_train.nunique()
}).sort_values(by='missing_ratio', ascending=False)
display(train_stats)
```

### Отбор признаков

Убираем часть признаков по следующим причинам:
- `line` - много уникальных значений, частичная утечка данных
- `model`, `graphicsModel` - много уникальных значений
- `allCoreBoostSpeed`, `perSocketMemBW`, `socketCount`, `PCIeLanes`, `thermalSolution`, `configTDP` - много пропусков

Проверим корреляцию Спирмена между оставшимися числовыми признаками

```python
temp_drop_cols = ['line', 'model', 'graphicsModel', 'allCoreBoostSpeed', 'perSocketMemBW', 'socketCount', 'PCIeLanes', 'thermalSolution', 'configTDP']  

cur_cols = [c for c in X_train.columns if c not in temp_drop_cols]
spearman_corr = X_train[cur_cols].select_dtypes('number').corr(method='spearman')


plt.figure(figsize=(14, 11))
sns.heatmap(spearman_corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1)
plt.title('Spearman Correlation (X_train)')
plt.tight_layout()
```

Хоть для классфикации допустимы признаки с мультиколлинеарностью, по матрице можно подтвердить часть структурных дублей, а именно `numCores` и `numThreads`, а также `launchDate` и `processTech`


Удаление `numThreads` обоснованно, тк он является производным от `numCores`, а связь `launchDate` и `processTech` по сути является отражением эпохи, и вдобавок `launchDate` содержит много пропусков, поэтому стоит отказаться от первого

Остальные признаки удалять сомнительно, так как они могут быть полезны для классификации, несмотря на корреляцию

```python
DROP_COLS = [
    'model', 'line', 'allCoreBoostSpeed', 'perSocketMemBW', 
    'socketCount', 'PCIeLanes', 'thermalSolution', 'configTDP', 
    'numThreads', 'launchDate', 'graphicsModel'
]

FEATURE_COLS = [col for col in X_train.columns if col not in DROP_COLS]

CAT_COLS = X_train[FEATURE_COLS].select_dtypes(exclude='number').columns.tolist()

NUM_COLS = [col for col in FEATURE_COLS if col not in CAT_COLS]

X_train = X_train[FEATURE_COLS].copy()
X_test = X_test[FEATURE_COLS].copy()
```

### Пайплайн


```python
num_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

#для группировки редких категорий, чтобы псевдопризнаки не разрастались
ENCODER_MIN_FREQ = 5 

cat_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(
        min_frequency=ENCODER_MIN_FREQ,
        handle_unknown='ignore',
        sparse_output=False)
    )
])

transformer = ColumnTransformer([
    ('num', num_pipeline, NUM_COLS),
    ('cat', cat_pipeline, CAT_COLS)
])

X_train_pr = transformer.fit_transform(X_train)
X_test_pr = transformer.transform(X_test)

transformer
```

### Обучение логистической регрессии


В качестве регуляризации используем L2, так как она более устойчива к мультиколлинеарности признаков, чем L1

Для подбора оптимального C используем кроссвалидацию GridSearch с 5 фолдами и метрикой F1

```python

#отдельный пайплайн, чтобы не было утечек данных в кроссвалидации из-за трансформера
lr = Pipeline([('transformer', transformer), ('classifier', LogisticRegression(max_iter=1000))])

lr_grid = GridSearchCV(
    lr,
    param_grid={
        'classifier__C': np.logspace(-3, 2, 6),
        # сравнение None и balanced
        'classifier__class_weight': [None, 'balanced']
    },
    scoring='f1',
    cv=5
)

lr_grid.fit(X_train, y_train)

lr_best = lr_grid.best_estimator_

print(f"Оптимальный C = {lr_grid.best_params_['classifier__C']}")

calculate_metrics(lr_best, 'Logistic Regression')
```

### Вывод

Логистическая регрессия показывает стабильное разделение классов на стандартизированных признаках. ROC-AUC близок к единице, те классы хорошо ранжируются линейной моделью на выбранных признаках. Ошибки распределены поровну между ложными срабатываниями и пропусками, перекоса в одну сторону нет. Сдвиг recall и precision можно объянить небольшим числом положителльных объектов в выборке по сравнению с отрицательными

### Бинарная классификация с помошью SVM

Для подбора оптимального C используем кроссвалидацию GridSearch с 5 фолдами и метрикой F1

```python

#отдельный пайплайн, чтобы не было утечек данных в кросс валидации из-за трансформера
svm = Pipeline([('transformer', transformer), ('classifier', SVC(kernel='linear'))])


svm_grid = GridSearchCV(
    svm,
    param_grid={
        'classifier__C': np.logspace(-3, 2, 6)
    },
    scoring='f1',
    cv=5
)

svm_grid.fit(X_train, y_train)

svm_best = svm_grid.best_estimator_

print(f"Оптимальный C = {svm_grid.best_params_['classifier__C']}")

calculate_metrics(svm_best, 'Linear SVM')
```

### Учет несбалансированности классов

Для учета несбалансированности классов можно использовать раздельные штрафы для положительного и отрицательного классов. Это реализуется с помощью `class_weight=balanced`, что заставляет модель автоматически перераспределять штрафы обратно пропорционально частотам классов


```python

svm_balanced = GridSearchCV(
    svm,
    param_grid={
        'classifier__C': np.logspace(-3, 2, 6),
        'classifier__class_weight': ['balanced']
    },
    scoring='f1',
    cv=5
)

svm_balanced.fit(X_train, y_train)

calculate_metrics(svm_balanced, 'Linear SVM (сбалансированный)')
```

### Вывод

Линейный SVM показывает качество, сопоставимое с логистической регрессией. ROC-AUC практически совпадает, а F1 и recall отличаются на пару объектов, так что можно сказать, что разница лежит в пределах шума.

Balanced SVM показывает рост recall до уровня, близкого к логистической регрессии, но за счет precision итоговоый F1 изменился незначительно, а ROC-AUC местами снизился. Это говорит о том, что изначальный дисбаланс классов умеренный, а балансировка в данном случае не сильно эффективна


### Сравнение моделей

```python
from sklearn.dummy import DummyClassifier

dummy = DummyClassifier(strategy='most_frequent')
dummy.fit(X_train_pr, y_train)

rows = [
    calculate_metrics(dummy, 'DummyClassifier', plots=False),
    calculate_metrics(lr_best, 'LogisticRegression', plots=False),
    calculate_metrics(svm_best, 'Linear SVM', plots=False),
    calculate_metrics(svm_balanced, 'Linear SVM balanced', plots=False),
]
display(pd.DataFrame(rows))

fig, ax = plt.subplots(1, 1, figsize=(6, 6))
RocCurveDisplay.from_estimator(lr_best, X_test, y_test, ax=ax, name='Logistic Regression')
RocCurveDisplay.from_estimator(svm_best, X_test, y_test, ax=ax, name='Linear SVM')
RocCurveDisplay.from_estimator(svm_balanced, X_test, y_test, ax=ax, name='Linear SVM balanced')
RocCurveDisplay.from_estimator(dummy, X_test, y_test, ax=ax, name='DummyClassifier')
```

Все три вариации модели значительно преввосходят dummy-классификатор по всем параметрам, что говорит о том, что построенные моделли способны выполнять задачу классификации удовлетворительно

### 3.4 В каких кейсах/доменах следует применять те или иные варианты? Запишите выводы

1. **macro** — когда даже узкий/редкий класс критически важен (лучше получить FP, чем пропустить редкий TP): медицина (диагностика редких заболеваний), системы антифрода, выявление дефектов на производстве.

4. **samples** — когда классовое распределение не важно, а критично оценить качество работы модели по каждому конкретному объекту индивидуально: задачи мультилейбл классификации (Multi-label classification), тегирование статей и категоризация мультимодального контента (распознавание нескольких объектов на фото).

## Задание 4. Выбросов и новизна. OneClassSVM

### 1. Теоретическая характеристика

необходимо уметь определять, принадлжеит ли новое наблюдение к тому же распределению, что и существующие

- выбросы - модели для обнаружения выбросов стараются соответствовать областям с наибольшей концентраицей обучающих данных. *неконтроллируемое обнаружение аномалий* - выбросы\аномалии не могут образовывать плотный кластер
- новизна - обучающеи данные не загрязнены и нам нужно определеить, является ли новое наблюдение выбросом, в данном контексте выброс называется **новизной**. *Полуконтроллируемое обнаружение аномалий* - новинки\аномалии могут образовывать плотный кластер в областях низкой плотности обучающих данных, которые в данном контексте считаются нормальными

`OneClassSVM` это алгоритм обучения без учителя, который описывает нормальные данные и отсекает аномалии. `SVM` проводит гиперплоскость, разделяющую классы, а `OneClassSVM`
- SupportVectorDataDescription гиперсфера минимального радииуса, содержащая почти все объекты выборки
- Шёлькопф - отображение данных в другое пространство и там выбирается гиперплоскость отделяющая все точки от начала координат с минимальным зазором
  - используется RBF-ядро
  - $\nu \in (0, 1]$ - верхняя граница доли выбросов
  - $\gamma > 0$ - обратная ширина RBF-ядра, то есть при увеличении параметра граница становится более локальной и извилистой
  - RBF ядро требует масштабирования признаков, так как основано на евклидовом расстоянии

### 2. Анализ выбросов в данных

```python
from sklearn.svm import OneClassSVM

# novelty detection
NU_OUTLIERS = 0.05

oc_svm = OneClassSVM(kernel='rbf', nu=NU_OUTLIERS, gamma='scale')
# unsupervised
oc_svm.fit(X_train_pr)

train_preds = oc_svm.predict(X_train_pr)
test_preds = oc_svm.predict(X_test_pr)

# distance between object and border
train_scores = oc_svm.decision_function(X_train_pr)

outliers_train_count = (train_preds == -1).sum()
outliers_train_ratio = outliers_train_count / len(train_preds) * 100

outliers_test_count = (test_preds == -1).sum()
outliers_test_ratio = outliers_test_count / len(test_preds) * 100

print(f"train size: {len(train_preds)}")
print(f"train outliers: {outliers_train_count} ({outliers_train_ratio:.2f}%)")

print("---")
print(f"test size: {len(test_preds)}")
print(f"test outliers: {outliers_test_count} ({outliers_test_ratio:.2f}%)")
```

```python
train_analysis_df = df.loc[X_train.index].copy()
train_analysis_df['anomaly'] = np.where(train_preds == -1, 'outline', 'inline')
train_analysis_df['score'] = train_scores

outliers_sample = train_analysis_df[train_analysis_df['anomaly'] == 'outline'].sort_values(by='score')
display(outliers_sample[['model', 'family', 'numCores', 'baseClock', 'defaultTDP', 'unlocked', 'score']].head(10))
```
